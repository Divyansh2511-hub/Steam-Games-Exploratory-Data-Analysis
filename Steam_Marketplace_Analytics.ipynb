# **7. Data Cleaning**

```python
df.dropna(subset=['Release_Date'], inplace=True)
print(f"Remaining rows after dropping missing Release_Date: {len(df)}")
```

Now, let's verify that all missing values have been addressed:

```python
# Convert 'Release_Date' to datetime objects
df['Release_Date'] = pd.to_datetime(df['Release_Date'], errors='coerce')

print("Data type of 'Release_Date' after conversion:")
display(df['Release_Date'].dtype)
```

Let's re-check the DataFrame information to confirm the change:

```python
df.info()
```

## **8.1 Analyze Release Trends**

### Subtask:
Extract the year from the 'Release_Date' column and visualize the number of game releases per year.


**Reasoning**:
Extracting the year from 'Release_Date', counting releases per year, and plotting the trend are all part of the current subtask's instructions. This step covers all remaining instructions.



```python
df['Release_Year'] = df['Release_Date'].dt.year

release_trends = df['Release_Year'].value_counts().sort_index()

plt.figure(figsize=(12, 6))
release_trends.plot(kind='line', marker='o')
plt.title('Game Releases Over Time')
plt.xlabel('Release Year')
plt.ylabel('Number of Games Released')
plt.grid(True)
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

## **8.2 Analyze Genre Distribution**

### Subtask:
Determine the distribution of 'Primary_Genre' and visualize the top genres using a bar chart to understand the most popular game categories.


**Reasoning**:
I will calculate the frequency of each 'Primary_Genre' and store the top genres to prepare for visualization as per the instructions.



```python
genre_distribution = df['Primary_Genre'].value_counts().head(10)
print("Top 10 Primary Game Genres:\n")
display(genre_distribution)
```

**Reasoning**:
I will create a bar chart to visualize the top 10 primary game genres, as calculated in the previous step, to understand their distribution and popularity.



```python
genre_distribution = df['Primary_Genre'].value_counts().head(10)
print("Top 10 Primary Game Genres:\n")
display(genre_distribution)

plt.figure(figsize=(12, 7))
sns.barplot(x=genre_distribution.index, y=genre_distribution.values, palette='viridis')
plt.title('Top 10 Primary Game Genres Distribution')
plt.xlabel('Primary Genre')
plt.ylabel('Number of Games')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

## **8.3 Analyze Pricing Patterns**

### Subtask:
Examine the distribution of 'Price_USD' using a histogram or box plot to identify common pricing tiers and potential outliers. Also, investigate how prices vary across different genres.


**Reasoning**:
I will start by plotting a histogram of the 'Price_USD' column to visualize its overall distribution and identify common pricing tiers, as per the first instruction of the subtask.



```python
plt.figure(figsize=(10, 6))
sns.histplot(df['Price_USD'], bins=30, kde=True)
plt.title('Distribution of Game Prices (USD)')
plt.xlabel('Price (USD)')
plt.ylabel('Number of Games')
plt.grid(True)
plt.tight_layout()
plt.show()
```

**Reasoning**:
Now that the overall distribution of prices has been visualized, I will create a box plot to show how 'Price_USD' varies across the top 10 'Primary_Genre' categories to identify differences in pricing strategies among genres, as per the subtask instructions.



```python
top_genres = df['Primary_Genre'].value_counts().head(10).index
df_top_genres = df[df['Primary_Genre'].isin(top_genres)]

plt.figure(figsize=(14, 7))
sns.boxplot(x='Primary_Genre', y='Price_USD', data=df_top_genres, palette='viridis')
plt.title('Game Prices Distribution Across Top 10 Genres')
plt.xlabel('Primary Genre')
plt.ylabel('Price (USD)')
plt.xticks(rotation=45, ha='right')
plt.grid(axis='y')
plt.tight_layout()
plt.show()
```

## **8.4 Analyze Review Score Distribution**

### Subtask:
Visualize the distribution of 'Review_Score_Pct' using a histogram to understand the overall sentiment towards games. Analyze if certain genres tend to have higher or lower review scores.


**Reasoning**:
I will plot a histogram of the 'Review_Score_Pct' column to visualize its overall distribution, as per the first instruction of the subtask.



```python
plt.figure(figsize=(10, 6))
sns.histplot(df['Review_Score_Pct'], bins=20, kde=True)
plt.title('Distribution of Game Review Scores')
plt.xlabel('Review Score Percentage')
plt.ylabel('Number of Games')
plt.grid(True)
plt.tight_layout()
plt.show()
```

## **8.5 Analyze Player Engagement**

### Subtask:
Investigate player engagement metrics such as 'Estimated_Owners' and '24h_Peak_Players'. Visualize their distributions and explore relationships with other variables like genre or review score.


**Reasoning**:
I will plot histograms for 'Estimated_Owners' and '24h_Peak_Players' to understand their distributions, as per the first instruction of the subtask.



```python
plt.figure(figsize=(15, 6))

plt.subplot(1, 2, 1)
sns.histplot(df['Estimated_Owners'], bins=50, kde=True)
plt.title('Distribution of Estimated Owners')
plt.xlabel('Estimated Owners')
plt.ylabel('Number of Games')
plt.grid(True)

plt.subplot(1, 2, 2)
sns.histplot(df['24h_Peak_Players'], bins=50, kde=True)
plt.title('Distribution of 24h Peak Players')
plt.xlabel('24h Peak Players')
plt.ylabel('Number of Games')
plt.grid(True)

plt.tight_layout()
plt.show()
```

**Reasoning**:
I will create box plots to visualize the distribution of 'Estimated_Owners' and '24h_Peak_Players' across the top 10 primary genres to understand how engagement metrics vary by genre. This addresses the 'explore relationships with other variables like genre' part of the subtask.



```python
top_genres = df['Primary_Genre'].value_counts().head(10).index
df_top_genres_engagement = df[df['Primary_Genre'].isin(top_genres)]

plt.figure(figsize=(18, 7))

plt.subplot(1, 2, 1)
sns.boxplot(x='Primary_Genre', y='Estimated_Owners', data=df_top_genres_engagement, palette='viridis')
plt.title('Estimated Owners Distribution Across Top 10 Genres')
plt.xlabel('Primary Genre')
plt.ylabel('Estimated Owners')
plt.xticks(rotation=45, ha='right')
plt.grid(axis='y')

plt.subplot(1, 2, 2)
sns.boxplot(x='Primary_Genre', y='24h_Peak_Players', data=df_top_genres_engagement, palette='viridis')
plt.title('24h Peak Players Distribution Across Top 10 Genres')
plt.xlabel('Primary Genre')
plt.ylabel('24h Peak Players')
plt.xticks(rotation=45, ha='right')
plt.grid(axis='y')

plt.tight_layout()
plt.show()
```

**Reasoning**:
I will create scatter plots to visualize the relationship between 'Estimated_Owners' and 'Review_Score_Pct', and between '24h_Peak_Players' and 'Review_Score_Pct'. This will allow us to investigate if there's a correlation between game engagement and review scores, as stated in the subtask.



```python
plt.figure(figsize=(15, 6))

plt.subplot(1, 2, 1)
sns.scatterplot(x='Review_Score_Pct', y='Estimated_Owners', data=df, alpha=0.6)
plt.title('Estimated Owners vs. Review Score Percentage')
plt.xlabel('Review Score Percentage')
plt.ylabel('Estimated Owners')
plt.grid(True)

plt.subplot(1, 2, 2)
sns.scatterplot(x='Review_Score_Pct', y='24h_Peak_Players', data=df, alpha=0.6)
plt.title('24h Peak Players vs. Review Score Percentage')
plt.xlabel('Review Score Percentage')
plt.ylabel('24h Peak Players')
plt.grid(True)

plt.tight_layout()
plt.show()
```

## **8.6 Correlation Analysis**

### Subtask:
Investigate the relationships between game `Price_USD`, `Review_Score_Pct`, `Estimated_Owners`, and `24h_Peak_Players` using scatter plots and a correlation heatmap.

**Reasoning**:
I will generate scatter plots to visualize the pairwise relationships between `Price_USD` vs `Review_Score_Pct`, `Review_Score_Pct` vs `Estimated_Owners`, and `Estimated_Owners` vs `24h_Peak_Players` to understand their individual correlations.

```python
plt.figure(figsize=(18, 6))

# Price vs Reviews
plt.subplot(1, 3, 1)
sns.scatterplot(x='Price_USD', y='Review_Score_Pct', data=df, alpha=0.6)
plt.title('Price vs. Review Score')
plt.xlabel('Price (USD)')
plt.ylabel('Review Score Percentage')
plt.grid(True)

# Reviews vs Owners
plt.subplot(1, 3, 2)
sns.scatterplot(x='Review_Score_Pct', y='Estimated_Owners', data=df, alpha=0.6)
plt.title('Review Score vs. Estimated Owners')
plt.xlabel('Review Score Percentage')
plt.ylabel('Estimated Owners')
plt.grid(True)

# Owners vs Peak Players
plt.subplot(1, 3, 3)
sns.scatterplot(x='Estimated_Owners', y='24h_Peak_Players', data=df, alpha=0.6)
plt.title('Estimated Owners vs. 24h Peak Players')
plt.xlabel('Estimated Owners')
plt.ylabel('24h Peak Players')
plt.grid(True)

plt.tight_layout()
plt.show()
```

**Reasoning**:
To get a comprehensive overview of the linear relationships between multiple numerical variables, I will calculate and visualize the correlation matrix using a heatmap.

```python
correlation_matrix = df[['Price_USD', 'Review_Score_Pct', 'Total_Reviews', 'Estimated_Owners', '24h_Peak_Players']].corr()

plt.figure(figsize=(10, 8))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation Matrix of Key Numerical Variables')
plt.tight_layout()
plt.show()
```

Identify unusual games by visualizing the distribution of key numerical features using box plots to detect outliers.

# **9. Key Findings - EDA Summary**

*   The platform has seen a massive surge in game releases, especially in recent years (peaking around 2025-2026 within this dataset), showing a booming market.

*  The 'Action' genre dominates the Steam marketplace by a significant margin, with 'Adventure' and 'Indie' also popular but much smaller.

*   Most games are either free or priced affordably, typically under $20-$30, making Steam accessible to many players. Pricing varies by genre, with 'Racing' and 'Simulation' often having slightly higher median prices.

*   The majority of games enjoy strong positive reviews (above 80% approval). However, there's a visible spread of games with very low or no reviews, indicating a diverse range of quality and reception.

*  Player numbers (estimated owners, peak players) are heavily skewed. While most games have a small audience, a few mega-hits grab a huge chunk of players, acting as significant outliers.

*   A game's price has almost no correlation with its review score or how many owners/players it attracts. A higher price doesn't guarantee better reviews or more engagement.

* Unsurprisingly, games with more reviews tend to have more estimated owners and higher peak player counts. These metrics strongly influence each other.

*  The dataset contains clear outliers: games that are unusually expensive, poorly reviewed, or have an exceptionally large player base. These point to unique market successes, failures, or specific niche strategies.
